## Problem 1 — FizzBuzz Extended (20 pts)
Print numbers from 1 to 100. For multiples of 3 print `Fizz`, multiples of 5 print `Buzz`, multiples of both print `FizzBuzz`, multiples of 7 print `Bang`, multiples of both 3 and 7 print `FizzBang`.

```python
# Problem 1 — FizzBuzz Extended
# Build the word piece by piece, so combinations come out automatically:
# 15 -> FizzBuzz, 21 -> FizzBang, 35 -> BuzzBang
for n in range(1, 101):
    word = ""
    if n % 3 == 0:
        word += "Fizz"
    if n % 5 == 0:
        word += "Buzz"
    if n % 7 == 0:
        word += "Bang"

    print(word if word else n)
```

## Problem 2 — Perfect Numbers (20 pts)
A perfect number equals the sum of its proper divisors (e.g. 6 = 1+2+3).
Find and print all perfect numbers between 1 and 10,000.

```python
# Problem 2 — Perfect Numbers
print("Perfect numbers between 1 and 10,000:")

for n in range(2, 10001):
    divisor_sum = 1          # 1 divides every number
    i = 2
    # Divisors come in pairs (i, n // i), so we only need to check up to sqrt(n)
    while i * i <= n:
        if n % i == 0:
            divisor_sum += i
            if i != n // i:  # don't count a square root twice
                divisor_sum += n // i
        i += 1

    if divisor_sum == n:
        print(n)
```

## Problem 3 — Pascal's Triangle (20 pts)
Print the first 8 rows of Pascal's Triangle.

```python
# Problem 3 — Pascal's Triangle
ROWS = 8

for row in range(ROWS):
    # Indent so the triangle is centred (each number takes 4 characters)
    print("  " * (ROWS - row - 1), end="")

    value = 1
    for k in range(row + 1):
        print(f"{value:4}", end="")
        # Next value in the row: C(row, k+1) = C(row, k) * (row - k) / (k + 1)
        value = value * (row - k) // (k + 1)
    print()
```

## Problem 4 — ATM Simulator (40 pts)
Simulate an ATM with a starting balance of `5000`. Use a `while` loop to display a menu:
1. Check Balance
2. Deposit
3. Withdraw
4. Exit

Requirements:
- Withdrawals must not exceed the current balance
- Deposits must be positive
- Loop until the user selects Exit
- Print a transaction summary at exit (total deposited, total withdrawn, final balance)

```python
# Problem 4 — ATM Simulator
balance = 5000.0
total_deposited = 0.0
total_withdrawn = 0.0

while True:
    print("\n===== ATM MENU =====")
    print("1. Check Balance")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Exit")
    choice = input("Choose an option (1-4): ").strip()

    if choice == "1":
        print(f"Current balance: {balance:.2f}")

    elif choice == "2":
        amount = float(input("Enter amount to deposit: "))
        if amount <= 0:
            print("Deposit must be a positive amount.")
        else:
            balance += amount
            total_deposited += amount
            print(f"Deposited {amount:.2f}. New balance: {balance:.2f}")

    elif choice == "3":
        amount = float(input("Enter amount to withdraw: "))
        if amount <= 0:
            print("Withdrawal must be a positive amount.")
        elif amount > balance:
            print(f"Insufficient funds. Your balance is {balance:.2f}")
        else:
            balance -= amount
            total_withdrawn += amount
            print(f"Withdrew {amount:.2f}. New balance: {balance:.2f}")

    elif choice == "4":
        break

    else:
        print("Invalid option. Please enter 1, 2, 3 or 4.")

print("\n===== TRANSACTION SUMMARY =====")
print(f"Total deposited: {total_deposited:.2f}")
print(f"Total withdrawn: {total_withdrawn:.2f}")
print(f"Final balance:   {balance:.2f}")
print("Thank you for using the ATM. Goodbye!")
```
