## Problem 1 — Arithmetic & Logic (25 pts)
Accept three integers from the user and print:
- Their sum, average, maximum, and minimum
- Whether all three are positive (`True`/`False`)
- Whether at least one is even (`True`/`False`)

```python
# Problem 1 — Arithmetic & Logic
num1 = int(input("Enter the first integer: "))
num2 = int(input("Enter the second integer: "))
num3 = int(input("Enter the third integer: "))

total = num1 + num2 + num3
average = total / 3

all_positive = num1 > 0 and num2 > 0 and num3 > 0
one_even = num1 % 2 == 0 or num2 % 2 == 0 or num3 % 2 == 0

print(f"Sum:     {total}")
print(f"Average: {average:.2f}")
print(f"Maximum: {max(num1, num2, num3)}")
print(f"Minimum: {min(num1, num2, num3)}")
print(f"All three positive: {all_positive}")
print(f"At least one even:  {one_even}")
```

## Problem 2 — String Manipulation (25 pts)
Accept a full name from the user (e.g. `"John Doe"`) and print:
- The name in uppercase and lowercase
- The number of characters (excluding spaces)
- The initials (first letter of each word, e.g. `J.D.`)
- Whether the name contains the letter `'a'` (case-insensitive)

```python
# Problem 2 — String Manipulation
full_name = input("Enter your full name: ").strip()

char_count = len(full_name.replace(" ", ""))
initials = ".".join(word[0].upper() for word in full_name.split()) + "."
has_a = "a" in full_name.lower()

print(f"Uppercase: {full_name.upper()}")
print(f"Lowercase: {full_name.lower()}")
print(f"Characters (excluding spaces): {char_count}")
print(f"Initials: {initials}")
print(f"Contains 'a': {has_a}")
```

## Problem 3 — BMI Calculator (25 pts)
Accept weight (kg) and height (m) from the user and calculate `BMI = weight / height²`.

Print the BMI rounded to 2 decimal places and its category:
- < 18.5 → Underweight
- 18.5–24.9 → Normal
- 25–29.9 → Overweight
- ≥ 30 → Obese

```python
# Problem 3 — BMI Calculator
weight = float(input("Enter your weight in kg: "))
height = float(input("Enter your height in m: "))

bmi = weight / height ** 2

# Using < 25 and < 30 (not <= 24.9 / <= 29.9) so values like 24.95 still get a category
if bmi < 18.5:
    category = "Underweight"
elif bmi < 25:
    category = "Normal"
elif bmi < 30:
    category = "Overweight"
else:
    category = "Obese"

print(f"BMI: {bmi:.2f}")
print(f"Category: {category}")
```

## Problem 4 — Swap Without Temp (25 pts)
Assign `a = 100`, `b = 250`. Swap their values **without** using a third variable. Print before and after.

```python
# Problem 4 — Swap Without Temp
a = 100
b = 250
print(f"Before swap: a = {a}, b = {b}")

# Arithmetic swap: no third variable needed
a = a + b   # a = 350
b = a - b   # b = 100
a = a - b   # a = 250

print(f"After swap:  a = {a}, b = {b}")

# Pythonic alternative (also no third variable): a, b = b, a
```
